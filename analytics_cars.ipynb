```python
import pandas as pd
import matplotlib.pyplot as plt

COLOR = "#F15B4E"

df = pd.read_csv("/content/CarsData.csv")
```

# Marcas com mais registros

```python
best_manufacturer = df['Manufacturer'].value_counts().sort_values(ascending=False)

plt.title('Marcas com mais registros')
plt.bar(best_manufacturer.index, best_manufacturer.values, color=COLOR)
plt.xticks(rotation=90)
plt.show()
```

# Evolução de lançamentos por ano

```python
df['year'].value_counts().sort_index().plot(
    kind='bar',
    color=COLOR,
    title='Evolução de lançamentos por ano',
    xlabel='Ano',
    ylabel='Quantidade'
  )
```

# Combustível mais comum

```python
df['fuelType'].value_counts().plot(
    kind='bar',
    color=COLOR,
    title='Combustível mais comum',
    xlabel='Combustível',
    ylabel='Quantidade'
  )
```

# Transmissão mais utilizada

```python
df['transmission'].value_counts().plot(
    kind='bar',
    color=COLOR,
    title='Transmissão mais utilizada',
    xlabel='Transmissão',
    ylabel='Quantidade'
  )
```

# Quantidade de tipo de transmissão por ano

```python

transmission = df[df['year'] >= 2010]
transmission = transmission.groupby('year')['transmission'].value_counts().unstack().plot(
    kind='bar',
    color=['#FFD700', '#C0C0C0', '#008000', '#0000FF'],
    title='Quantidade de tipo de transmissão por ano',
    xlabel='Ano',
    ylabel='Quantidade'
  )
```
